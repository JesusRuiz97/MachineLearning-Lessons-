# Solucion:

```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier  
def generate_dataset(n_samples=100, seed=42):
    np.random.seed(seed)
    
    X = np.random.rand(n_samples, 5)
    
    X[:, 3] = np.random.randint(0, 3, n_samples)
    
    y = []
    
    for i in range(n_samples):
        mantenimiento = X[i, 1]
        libs = X[i, 2]
        tipo_app = X[i, 3]
        rendimiento = X[i, 4]
        
        if rendimiento > 0.8 and tipo_app == 2:
            lenguaje = 3  # C++
        elif mantenimiento > 0.7 and tipo_app == 1:
            lenguaje = 2  # Java
        elif libs > 0.6 and tipo_app == 0:
            lenguaje = 0  # Python
        else:
            lenguaje = 1  # JavaScript
            
        y.append(lenguaje)
        
    return X, np.array(y)

class LanguagePredictor:
    def __init__(self):
        self.model = RandomForestClassifier(n_estimators=100, random_state=42)
        
        self.label_map = {
            0: "Python",
            1: "JavaScript",
            2: "Java",
            3: "C++"
        }
        
    def train(self, X, y):
        self.model.fit(X, y)
        
    def predict(self, features):
        features = np.array(features)
        if features.ndim == 1:
            features = features.reshape(1, -1)
            
        prediction_id = self.model.predict(features)[0]
        return self.label_map.get(prediction_id, "Desconocido")
```
