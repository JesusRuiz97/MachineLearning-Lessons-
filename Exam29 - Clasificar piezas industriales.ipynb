# Solucion:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

class Piece:
    def __init__(self, texture, symmetry, edges, center_offset, label):
        self.texture = texture
        self.symmetry = symmetry
        self.edges = edges
        self.center_offset = center_offset
        self.label = label

    def to_vector(self) -> list:

        return [self.texture, self.symmetry, self.edges, self.center_offset]

class PieceDatasetGenerator:
    def __init__(self, n=400):
        self.n = n

    def generate(self) -> list:
        np.random.seed(42)
        pieces = []

        textures = np.random.normal(loc=0.5, scale=0.15, size=self.n)
        symmetries = np.random.normal(loc=0.5, scale=0.15, size=self.n)
        edges_list = np.random.normal(loc=45, scale=10, size=self.n)
        offsets = np.random.normal(loc=0.2, scale=0.1, size=self.n)

        for i in range(self.n):
            t = max(0.0, min(1.0, textures[i]))
            s = max(0.0, min(1.0, symmetries[i]))
            e = max(0.0, edges_list[i])
            o = max(0.0, offsets[i])
            
            label = "Correcta"
            
            cond1 = (s < 0.4) and (o > 0.25)
            cond2 = (t < 0.35)
            cond3 = (e < 30)
            cond4 = (o > 0.35)

            if cond1 or cond2 or cond3 or cond4:
                label = "Defectuosa"
            
            pieces.append(Piece(t, s, e, o, label))
            
        return pieces

class PieceClassifier:
    def __init__(self):
        self.model = SVC(kernel='rbf', gamma='scale', C=1.0)

    def fit(self, pieces: list) -> None:
        X = [p.to_vector() for p in pieces]
        y = [p.label for p in pieces]
        self.model.fit(X, y)

    def predict(self, texture, symmetry, edges, center_offset) -> str:
        features = [[texture, symmetry, edges, center_offset]]
        return self.model.predict(features)[0]

    def evaluate(self, test_data: list) -> None:
        X_test = [p.to_vector() for p in test_data]
        y_test = [p.label for p in test_data]
        
        y_pred = self.model.predict(X_test)
        
        print("📊 Matriz de confusión:")
        print(confusion_matrix(y_test, y_pred))
        print("\n📝 Informe de clasificación:")
        print(classification_report(y_test, y_pred))

class PieceAnalysisExample:
    def run(self) -> None:
        print("🏭 Generando piezas industriales...")
        generator = PieceDatasetGenerator(n=400)
        dataset = generator.generate()
        
        train_data, test_data = train_test_split(dataset, test_size=0.3, random_state=42)
        
        print("🤖 Entrenando modelo de IA (SVM)...")
        classifier = PieceClassifier()
        classifier.fit(train_data)
        
        classifier.evaluate(test_data)
        
        custom_feat = (0.45, 0.5, 45, 0.15)
        resultado = classifier.predict(*custom_feat)
        
        print("\n🔎 Predicción de pieza personalizada:")
        print(f"  → Textura: {custom_feat[0]}, Simetría: {custom_feat[1]}, Bordes: {custom_feat[2]}, Offset: {custom_feat[3]}")
        print(f"  → Clasificación: {resultado}")
        
        self.visualize_results(dataset)

    def visualize_results(self, pieces):
        data = []
        for p in pieces:
            data.append({
                "Textura": p.texture,
                "Simetría": p.symmetry,
                "Bordes": p.edges,
                "Offset": p.center_offset,
                "Etiqueta": p.label
            })
        
        df = pd.DataFrame(data)
        
        plt.figure(figsize=(10, 6))
        
        colors = {'Correcta': 'green', 'Defectuosa': 'red'}
        
        for label, color in colors.items():
            subset = df[df['Etiqueta'] == label]
            plt.scatter(subset['Textura'], subset['Offset'], 
                        c=color, label=label, alpha=0.6, edgecolors='k')

        plt.title("🏭 Clasificación de piezas industriales")
        plt.xlabel("Textura")
        plt.ylabel("Offset (Desviación)")
        plt.legend()
        plt.grid(True, linestyle='--', alpha=0.7)
        
        plt.show()

if __name__ == "__main__":
    example = PieceAnalysisExample()
    example.run()
```
