# Solucion:

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_iris

def knn_clasificacion(datos, k=3):
    """
    Entrena un modelo KNN con los datos proporcionados.
    """

    X = datos[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
    

    y = datos['species']
    

    knn = KNeighborsClassifier(n_neighbors=k)
    

    knn.fit(X, y)
    

    return knn



iris = load_iris(as_frame=True)
data = iris.frame

data.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

modelo_knn = knn_clasificacion(data, k=3)

nuevas_muestras = pd.DataFrame({
    'sepal_length': [5.1, 6.0, 4.4],
    'sepal_width': [3.5, 2.9, 3.2],
    'petal_length': [1.4, 4.5, 1.3],
    'petal_width': [0.2, 1.5, 0.2]
})

estimaciones = modelo_knn.predict(nuevas_muestras)

nombres_especies = iris.target_names
predicciones_nombres = [nombres_especies[i] for i in estimaciones]

print("Estimaciones de Clasificación:")
print(predicciones_nombres)
```
