# Solucion:

```python
import numpy as np
from sklearn.cluster import KMeans
from typing import List

class Player:
    def __init__(self, name: str, avg_session_time: float, missions_completed: int, accuracy: float, aggressiveness: float):
        self.name = name
        self.avg_session_time = avg_session_time
        self.missions_completed = missions_completed
        self.accuracy = accuracy
        self.aggressiveness = aggressiveness

    def to_vector(self) -> List[float]:
        return [self.avg_session_time, self.missions_completed, self.accuracy, self.aggressiveness]

    def __repr__(self):
        return self.name

class PlayerClusterer:
    def __init__(self):
        self.model = None

    def fit(self, players: List[Player], n_clusters: int):
        data = [p.to_vector() for p in players]
        self.model = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
        self.model.fit(data)

    def predict(self, player: Player) -> int:
        if self.model is None:
            raise Exception("El modelo no ha sido entrenado aún.")
        
        data = [player.to_vector()]

        return int(self.model.predict(data)[0])

    def get_cluster_centers(self):
        return self.model.cluster_centers_

    def print_cluster_summary(self, players: List[Player]):
        if self.model is None:
            return

        labels = self.model.labels_
        cluster_map = {}
        
        for player, label in zip(players, labels):

            label = int(label) 
            if label not in cluster_map:
                cluster_map[label] = []
            cluster_map[label].append(player.name)

        for label in sorted(cluster_map.keys()):
            print(f"Cluster {label}:")
            for name in cluster_map[label]:
                print(f"  - {name}")

class GameAnalytics:
    def run(self):
        raw_data = [
            ("Alice", 2.5, 100, 0.85, 0.3),
            ("Bob", 1.0, 20, 0.60, 0.7),
            ("Charlie", 3.0, 150, 0.9, 0.2),
            ("Diana", 0.8, 15, 0.55, 0.9),
            ("Eve", 2.7, 120, 0.88, 0.25),
            ("Frank", 1.1, 30, 0.62, 0.65),
            ("Grace", 0.9, 18, 0.58, 0.85),
            ("Hank", 3.2, 160, 0.91, 0.15)
        ]
```
