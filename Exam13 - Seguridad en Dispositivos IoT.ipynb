Contexto
Imagina que formas parte del equipo de defensa digital de un hogar inteligente. Con cada vez más dispositivos conectados a internet, se vuelve crucial detectar cuáles son seguros y cuáles podrían representar una amenaza.

Tu tarea será construir un modelo que aprenda a clasificar dispositivos IoT (Internet de las Cosas) como seguros (1) o peligrosos (0) según su tráfico de red, utilizando el algoritmo k-Nearest Neighbors (KNN).



🧠 Instrucciones

Crea una clase llamada IoTKNNClassifier que simule un sistema de detección para dispositivos IoT basado en aprendizaje automático.



🔧 1. Constructor: __init__(self, n_neighbors=3, n_samples=50)

# Solucion:

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

class IoTKNNClassifier:
    def __init__(self, n_neighbors=3, n_samples=50):
        self.n_neighbors = n_neighbors

        paquetes = np.random.randint(10, 1001, n_samples)
        bytes_paq = np.random.randint(50, 1501, n_samples)
        protocolo = np.random.randint(1, 4, n_samples)

        etiquetas = []
        for p, b in zip(paquetes, bytes_paq):
            if p > 500 or b > 800:
                etiquetas.append(0)
            else:
                etiquetas.append(1)
        
        data = {
            'paquetes_por_segundo': paquetes,
            'bytes_por_paquete': bytes_paq,
            'protocolo': protocolo,
            'seguro': etiquetas
        }
        
        self.df = pd.DataFrame(data)
        
        self.X = self.df.drop(columns=["seguro"])
        self.y = self.df["seguro"]
        
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=0.2, random_state=42
        )
        
        self.model = None

    def train(self):
        self.model = KNeighborsClassifier(n_neighbors=self.n_neighbors)
        self.model.fit(self.X_train, self.y_train)

    def evaluate(self):
        if self.model is None:
            return 0.0
        y_pred = self.model.predict(self.X_test)
        return accuracy_score(self.y_test, y_pred)

    def predict(self, nuevo_dispositivo):
        if self.model is None:
            self.train()
            
        prediction = self.model.predict([nuevo_dispositivo])[0]
        return prediction

clasificador = IoTKNNClassifier(n_neighbors=3, n_samples=100) # Aumenté un poco las muestras para estabilidad
clasificador.train()
precision = clasificador.evaluate()
print(f"Precisión del modelo: {precision}")

if precision > 0.5:
    print("🚀 ¡Prueba superada! El modelo ha aprendido el patrón.")
else:
    print("⚠️ La precisión sigue siendo baja.")
```
